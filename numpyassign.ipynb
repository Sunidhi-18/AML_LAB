```python
import numpy as np

marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])

# Marks greater than 75
print("Marks > 75:", marks[marks > 75])

# Marks between 50 and 80
print("Marks between 50 and 80:", marks[(marks >= 50) & (marks <= 80)])

# Number of students who scored below 50
print("Students below 50:", np.sum(marks < 50))

# Percentage of students who scored 60 or above
pct_60_plus = np.mean(marks >= 60) * 100
print(f"Percentage scoring 60 or above: {pct_60_plus}%")
```

```python
marks = np.array([
    [80, 75, 90],
    [65, 70, 72],
    [88, 92, 85],
    [55, 60, 58]
])

# Total & average marks of each student (row-wise)
student_totals = np.sum(marks, axis=1)
student_averages = np.mean(marks, axis=1)
print("Total marks of each student:", student_totals)
print("Average marks of each student:", student_averages)

# Average & highest mark in each subject (column-wise)
subject_averages = np.mean(marks, axis=0)
subject_max = np.max(marks, axis=0)
print("Average marks in each subject:", subject_averages)
print("Highest mark in each subject:", subject_max)

# Student with highest total (1-based index)
highest_student = np.argmax(student_totals) + 1
print("Student with highest total: Student", highest_student)
```

11. Create:

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

Perform:

Matrix addition
Element-wise multiplication
Matrix multiplication
Transpose of A.

explain the difference between:
A * B and A @ B

```python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

# Operations
add_res = A + B
elem_mult = A * B
matmul_res = A @ B  # or np.dot(A, B)
trans_A = A.T

print("Matrix Addition (A + B):\n", add_res)
print("\nElement-wise Multiplication (A * B):\n", elem_mult)
print("\nMatrix Multiplication (A @ B):\n", matmul_res)
print("\nTranspose of A (A.T):\n", trans_A)
```

```python
temperature = np.array([
    31, 32, 33, 35, 36, 34, 32,
    30, 29, 31, 33, 35, 37, 36
])

avg_temp = np.mean(temperature)
print("Mean temperature:", avg_temp)
print("Max temp:", np.max(temperature), "| Min temp:", np.min(temperature))
print("Days above average temp:", np.sum(temperature > avg_temp))
print("Temperatures > 34°C:", temperature[temperature > 34])
print("Difference between max & min:", np.max(temperature) - np.min(temperature))
print("Standard deviation:", np.std(temperature))
```

```python
sales = np.array([
    12000, 15000, 17000, 14000,
    18000, 21000, 19000, 22000,
    25000, 23000, 26000, 28000
])

print("Total annual sales:", np.sum(sales))
print("Average monthly sales:", np.mean(sales))
print("Month with highest sales: Month", np.argmax(sales) + 1)
print("Month with lowest sales: Month", np.argmin(sales) + 1)
print("Months with sales > ₹20,000:", np.sum(sales > 20000))

# Percentage increase from Jan (index 0) to Dec (index -1)
pct_increase = ((sales[-1] - sales[0]) / sales[0]) * 100
print(f"Percentage increase from Jan to Dec: {pct_increase:.2f}%")
```

```python
import numpy as np

marks = np.array([
    [78, 85, 90, 88],
    [65, 70, 72, 68],
    [92, 88, 95, 91],
    [55, 60, 58, 62],
    [80, 75, 82, 79]
])

student_totals = np.sum(marks, axis=1)
student_averages = np.mean(marks, axis=1)
subject_averages = np.mean(marks, axis=0)

print("Total marks of every student:", student_totals)
print("Average of every student:", student_averages)
print("Average mark for every subject:", subject_averages)
print("Highest-scoring student: Student", np.argmax(student_totals) + 1)
print("Lowest-scoring student: Student", np.argmin(student_totals) + 1)
print("Students with avg > 80: Students", np.where(student_averages > 80)[0] + 1)
print("Overall class average:", np.mean(marks))

# Categorize performance using np.select with explicit string default
conditions = [
    student_averages >= 85,
    (student_averages >= 70) & (student_averages < 85),
    (student_averages >= 50) & (student_averages < 70),
    student_averages < 50
]
categories = ['Excellent', 'Good', 'Average', 'Needs Improvement']

# Pass default='' to match string dtype
performance = np.select(conditions, categories, default='')

for i, cat in enumerate(performance):
    print(f"Student {i+1}: {cat}")
```

```python
# Formula: (X - min) / (max - min)
arr_norm = np.array([10, 25, 40, 55, 70, 85, 100], dtype=float)
normalized = (arr_norm - np.min(arr_norm)) / (np.max(arr_norm) - np.min(arr_norm))
print("Original Array:", arr_norm)
print("Normalized Array (0 to 1 scale):", normalized)
```

```python
# Formula: (X - mean) / std
arr_std = np.array([150, 160, 165, 170, 175, 180, 190], dtype=float)
standardized = (arr_std - np.mean(arr_std)) / np.std(arr_std)
print("Original Array:", arr_std)
print("Standardized Array (Mean=0, Std=1):", standardized)
```

```python
import time

size = 1_000_000

# Python List Multiplication
py_list = list(range(1, size + 1))
start_time = time.time()
py_result = [x * 2 for x in py_list]
py_time = time.time() - start_time

# NumPy Array Multiplication
np_array = np.arange(1, size + 1)
start_time = time.time()
np_result = np_array * 2
np_time = time.time() - start_time

print(f"Python List Execution Time: {py_time:.6f} seconds")
print(f"NumPy Array Execution Time: {np_time:.6f} seconds")
print(f"NumPy is {py_time / np_time:.2f}x faster!")
```
