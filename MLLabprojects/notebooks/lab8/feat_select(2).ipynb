For data-1.csv

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Load Data 1
df1 = pd.read_csv('C:/Users/MSIS/Downloads/AML/MLLabprojects/data/data-1.csv')
print("Data 1 Shape:", df1.shape)

# 2. Drop Constant Features (Zero Variance)
non_constant_cols1 = [col for col in df1.columns if df1[col].nunique() > 1]
constant_cols1 = list(set(df1.columns) - set(non_constant_cols1))

print(f"Dropped Constant Features: {constant_cols1}")
df1_clean = df1[non_constant_cols1]

# 3. Calculate Correlation Matrix
corr1 = df1_clean.corr()

# 4. Plot Heatmap
plt.figure(figsize=(6, 4))
sns.heatmap(corr1, annot=True, cmap='coolwarm', fmt=".2f", vmin=-1, vmax=1, linewidths=0.8)
plt.title('Correlation Heatmap - data-1.csv', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()

# 5. Feature Selection based on Target Correlation ('items')
target_col1 = 'items'
if target_col1 in corr1.columns:
    print(f"Correlation with Target ('{target_col1}'):")
    target_corr1 = corr1[target_col1].drop(target_col1)
    print(target_corr1)
    
    # Select features that are not the target column
    selected_features1 = [col for col in df1_clean.columns if col != target_col1]
    print(f"\nFinal Selected Features for Data 1: {selected_features1}")
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Load Data 2
df2 = pd.read_csv('C:/Users/MSIS/Downloads/AML/MLLabprojects/data/data-2.csv')
print("Data 2 Shape:", df2.shape)

# 2. Drop Constant Features (Zero Variance)
non_constant_cols2 = [col for col in df2.columns if df2[col].nunique() > 1]
constant_cols2 = list(set(df2.columns) - set(non_constant_cols2))

print(f"Dropped Constant Features: {constant_cols2}")
df2_clean = df2[non_constant_cols2]

# 3. Calculate Correlation Matrix
corr2 = df2_clean.corr()

# 4. Plot Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr2, annot=True, cmap='coolwarm', fmt=".2f", vmin=-1, vmax=1, linewidths=0.8)
plt.title('Correlation Heatmap - data-2.csv', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()

# 5. Feature Selection based on Target Correlation
# Replace 'target_column_name' with your actual target variable in data-2.csv
target_col2 = df2_clean.columns[-1]  # Defaults to the last column

print(f"Correlation with Target ('{target_col2}'):")
target_corr2 = corr2[target_col2].drop(target_col2)
print(target_corr2)

# Select non-target features (Optionally apply a correlation threshold e.g., > 0.1)
selected_features2 = [col for col in df2_clean.columns if col != target_col2]
print(f"\nFinal Selected Features for Data 2: {selected_features2}")
```
