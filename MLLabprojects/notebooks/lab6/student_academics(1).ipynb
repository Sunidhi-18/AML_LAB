Import & Data loading

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import Lasso, LassoCV, LinearRegression, Ridge, RidgeCV
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Create the dataset
data = {
    "Hours_Studied": [2, 4, 6, 8, 10],
    "Classes_attended": [3, 4, 5, 6, 8],
    "Final_Score": [65, 70, 75, 85, 95],
}

df = pd.DataFrame(data)

x = df[["Hours_Studied", "Classes_attended"]]
y = df["Final_Score"]

# New student data for prediction (7 hrs studied, 5 classes attended)
new_student = pd.DataFrame({"Hours_Studied": [7], "Classes_attended": [5]})
```

Define evaluation functions

```python
# Function to calculate all 3 regression error metrics
def calculate_errors(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)
    return {"MAE": mae, "MSE": mse, "RMSE": rmse, "R2": r2}
```

Multiple Linear Regression

```python
# Train OLS Model
ols = LinearRegression()
ols.fit(x, y)

# Predict on training data and new student
y_pred_ols = ols.predict(x)
ols_pred_new = ols.predict(new_student)[0]
ols_errors = calculate_errors(y, y_pred_ols)

print(f"OLS Intercept: {ols.intercept_:.4f}")
print(f"OLS Coefficients: {ols.coef_}")
print(f"OLS Prediction for New Student: {ols_pred_new:.2f}")
print("OLS Evaluation Errors:", ols_errors)
# Train OLS Model
ols = LinearRegression()
ols.fit(x, y)

# Predict on training data and new student
y_pred_ols = ols.predict(x)
ols_pred_new = ols.predict(new_student)[0]
ols_errors = calculate_errors(y, y_pred_ols)

print(f"OLS Intercept: {ols.intercept_:.4f}")
print(f"OLS Coefficients: {ols.coef_}")
print(f"OLS Prediction for New Student: {ols_pred_new:.2f}")
print("OLS Evaluation Errors:", ols_errors)
```

Ridge Regression (fixed alpha = 1.0)

```python
# Train Ridge Model with default alpha=1.0
ridge = Ridge(alpha=1.0)
ridge.fit(x, y)

# Predict on training data and new student
y_pred_ridge = ridge.predict(x)
ridge_pred_new = ridge.predict(new_student)[0]
ridge_errors = calculate_errors(y, y_pred_ridge)

print(f"Ridge (alpha=1.0) Prediction for New Student: {ridge_pred_new:.2f}")
print("Ridge Evaluation Errors:", ridge_errors)
```

RidgeCV (Optimal Alpha Tuning)

```python
# Define candidate alpha values
alpha_range = np.logspace(-3, 3, 100)

# Train RidgeCV model using Leave-One-Out Cross-Validation
ridge_cv = RidgeCV(alphas=alpha_range, cv=None)
ridge_cv.fit(x, y)

# Retrieve best alpha and evaluate
best_alpha_ridge = ridge_cv.alpha_
y_pred_ridge_cv = ridge_cv.predict(x)
ridge_cv_pred_new = ridge_cv.predict(new_student)[0]
ridge_cv_errors = calculate_errors(y, y_pred_ridge_cv)

print(f"Best Alpha for Ridge: {best_alpha_ridge:.4f}")
print(
    f"RidgeCV Prediction for New Student: {ridge_cv_pred_new:.2f}"
)
print("RidgeCV Evaluation Errors:", ridge_cv_errors)
```

Lasso Regression

```python
# Train Lasso Model with fixed alpha=1.0
lasso = Lasso(alpha=1.0)
lasso.fit(x, y)

# Predict on training data and new student
y_pred_lasso = lasso.predict(x)
lasso_pred_new = lasso.predict(new_student)[0]
lasso_errors = calculate_errors(y, y_pred_lasso)

print(f"Lasso (alpha=1.0) Prediction for New Student: {lasso_pred_new:.2f}")
print("Lasso Evaluation Errors:", lasso_errors)
```

LassoCV (Optimal Alpha Tuning)

```python
# Train LassoCV model using 5-fold Cross-Validation
lasso_cv = LassoCV(alphas=alpha_range, cv=5)
lasso_cv.fit(x, y)

# Retrieve best alpha and evaluate
best_alpha_lasso = lasso_cv.alpha_
y_pred_lasso_cv = lasso_cv.predict(x)
lasso_cv_pred_new = lasso_cv.predict(new_student)[0]
lasso_cv_errors = calculate_errors(y, y_pred_lasso_cv)

print(f"Best Alpha for Lasso: {best_alpha_lasso:.4f}")
print(
    f"LassoCV Prediction for New Student: {lasso_cv_pred_new:.2f}"
)
print("LassoCV Evaluation Errors:", lasso_cv_errors)
```
