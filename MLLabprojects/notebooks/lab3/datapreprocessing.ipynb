Taking care of missing data

```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer

# 1. Load your dataset
df = pd.read_csv('../data/data.csv')

# 2. Convert the dataframe into a 2D NumPy array 
# (Make sure you don't select just a single column, which makes it 1D)
x = df.values  

# 3. Now run your imputer code safely
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
imputer = imputer.fit(x[:, 1:3])
x[:, 1:3] = imputer.transform(x[:, 1:3])

# 4. View the result
print(x)
```

```python
#encoding categorical data
#encoding the independent variable
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
labelencoder_x = LabelEncoder()
x[:,0] = labelencoder_x.fit_transform(x[:,0])
print(x)
```

```python
#endoing the independent variable
from sklearn.compose import ColumnTransformer
ct = ColumnTransformer([("Country", OneHotEncoder(),[0])], remainder = "passthrough")
x = ct.fit_transform(x)

x.astype(int)
```

```python
#split data into training and test set
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x, y, test_size=0.2, random_state=45)
```

```python
#feature scaling
#sc_x = (x[1] - mean) / standard deviation
from sklearn.preprocessing import StandardScaler

sc_x = StandardScaler()

# Only scale columns 1 and 2 (leave the text columns untouched)
x_train[:, 1:3] = sc_x.fit_transform(x_train[:, 1:3])
x_test[:, 1:3] = sc_x.transform(x_test[:, 1:3])
```
