```python
#example for class
class Person:
  institute="Manipal school of Information science"# a class variable

  #self keyword to access the atribute and methods of the class

  def set_name(self,new_name): #a Method
    self.name=new_name
  def set_location(self,new_location):
    self.location=new_location
```

```python
person=Person()
person.set_name('Sunidhi')
person.set_location('Mangaluru,karnataka')
```

```python
print('{} lives in {} and studying in the institute {}'.format(person.name,person.location,person.institute))
```

```python
#example for the mapping function between the three lists

store1=[10.00,11.00,12.34,5.34]
store2=[19.00,10,10,12.34,5.01]
store3=[15.00,12.00,11.00,6.00]
cheapest=map(min,store1,store2,store3)
cheapest
```

```python
# Original list of numbers
numbers = [1, 2, 3, 4, 5]

# Apply lambda function to square each number
squared_numbers = list(map(lambda x: x ** 2, numbers))

print(squared_numbers)
# Output: [1, 4, 9, 16, 25]
```

```python
# Define two lists of numbers
list1 = [1, 2, 3, 4]
list2 = [10, 20, 30, 40]

# Use map() with a two-argument lambda function
# The lambda takes 'x' from list1 and 'y' from list2 simultaneously
result_map = map(lambda x, y: x + y, list1, list2)

# Convert the map object back into a list to see the results
result_list = list(result_map)

print(result_list)
# Output: [11, 22, 33, 44]
```

```python
people = ['python programing language', 'computer lab test', 'ram scored 100', 'sam scored 99']

def split_title_and_name(person):
    words = person.split()
    title = words[0]       # First word
    lastname = words[-1]   # Last word
    return f"{title} {lastname}"

# Map function over list
list(map(split_title_and_name, people))
```
