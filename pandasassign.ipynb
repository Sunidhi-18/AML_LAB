9. Add a new column called Result. Students scoring 75 or above should receive "Pass", otherwise "Fail".

```python
import numpy as np
import pandas as pd

# Recreating the DataFrame from previous steps
data = {
    'Name': ['Asha', 'Ravi', 'Meena', 'Kiran', 'Arun'],
    'Age': [22, 23, 21, 24, 22],
    'Department': ['CSE', 'ECE', 'CSE', 'ME', 'ECE'],
    'Marks': [85, 78, 92, 74, 88]
}

df = pd.DataFrame(data)

# Add the new 'Result' column using np.where(condition, if_true, if_false)
df['Result'] = np.where(df['Marks'] >= 75, 'Pass', 'Fail')

# Display the updated DataFrame
print("--- DataFrame with Result Column ---")
print(df)
```


10. Assign grades according to:

	| Marks | Grade |
	| ----: | ----- |
	|  ≥ 90 | A     |
	|  ≥ 80 | B     |
	|  ≥ 70 | C     |
	|  < 70 | D     |

```python
import numpy as np
import pandas as pd

data = {
    'Name': ['Asha', 'Ravi', 'Meena', 'Kiran', 'Arun'],
    'Age': [22, 23, 21, 24, 22],
    'Department': ['CSE', 'ECE', 'CSE', 'ME', 'ECE'],
    'Marks': [85, 78, 92, 74, 88]
}
df = pd.DataFrame(data)

conditions = [
    df['Marks'] >= 90,
    df['Marks'] >= 80,
    df['Marks'] >= 70,
    df['Marks'] < 70
]
choices = ['A', 'B', 'C', 'D']

# Pass a string to default to match the string choices
df['Grade'] = np.select(conditions, choices, default='D')

print(df)
```

13. Replace:

	Missing Age with average Age
	Missing Marks with average Marks

```python
import numpy as np
import pandas as pd

# Recreating the DataFrame from the previous step
data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun"],
    "Age": [22, 23, np.nan, 24, 22],
    "Marks": [85, np.nan, 92, 74, 88],
}

students = pd.DataFrame(data)

# Calculate the mean (average) for Age and Marks
avg_age = students['Age'].mean()
avg_marks = students['Marks'].mean()

# Replace missing values using fillna()
students['Age'] = students['Age'].fillna(avg_age)
students['Marks'] = students['Marks'].fillna(avg_marks)

# Display the updated DataFrame
print("--- DataFrame After Replacing Missing Values ---")
print(students)
```

Perform the following analysis:

Display the first five records.
Find the dimensions of the dataset.
Find the average final mark.
Find the highest-scoring student.
Find the lowest-scoring student.
Display students scoring above 80.
Find average marks by department.
Find students having attendance above 80%.
Sort students by FinalMarks.
Add a Result column: Pass if FinalMarks ≥ 70, otherwise Fail.
Add a performance category:
	≥90 → Excellent
	≥80 → Very Good
	≥70 → Good

Find the department having the highest average FinalMarks.

```python
import numpy as np
import pandas as pd

# Creating the dataset
data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun", "Divya", "John", "Priya"],
    "Department": ["CSE", "ECE", "CSE", "ME", "ECE", "CSE", "ME", "ECE"],
    "Attendance": [92, 75, 96, 68, 82, 88, 72, 95],
    "Internal": [85, 72, 94, 65, 80, 86, 70, 91],
    "FinalMarks": [88, 74, 96, 62, 83, 89, 68, 93],
}

df = pd.DataFrame(data)

# 1. Display the first five records
print("--- 1. First Five Records ---")
print(df.head(5))

# 2. Find the dimensions of the dataset
print("\n--- 2. Dimensions of the Dataset ---")
print(df.shape)  # (Rows, Columns)

# 3. Find the average final mark
print("\n--- 3. Average Final Mark ---")
print(df["FinalMarks"].mean())

# 4. Find the highest-scoring student
print("\n--- 4. Highest-Scoring Student ---")
highest_student = df.loc[df["FinalMarks"].idxmax()]
print(highest_student)

# 5. Find the lowest-scoring student
print("\n--- 5. Lowest-Scoring Student ---")
lowest_student = df.loc[df["FinalMarks"].idxmin()]
print(lowest_student)

# 6. Display students scoring above 80
print("\n--- 6. Students Scoring Above 80 ---")
print(df[df["FinalMarks"] > 80])

# 7. Find average marks by department
print("\n--- 7. Average Final Marks by Department ---")
print(df.groupby("Department")["FinalMarks"].mean())

# 8. Find students having attendance above 80%
print("\n--- 8. Students with Attendance Above 80% ---")
print(df[df["Attendance"] > 80])

# 9. Sort students by FinalMarks (descending order)
print("\n--- 9. Students Sorted by FinalMarks (Descending) ---")
print(df.sort_values(by="FinalMarks", ascending=False))

# 10. Add a Result column
df["Result"] = np.where(df["FinalMarks"] >= 70, "Pass", "Fail")

# 11. Add a performance category column
conditions = [
    df["FinalMarks"] >= 90,
    df["FinalMarks"] >= 80,
    df["FinalMarks"] >= 70,
    df["FinalMarks"] < 70,
]
categories = ["Excellent", "Very Good", "Good", "Needs Improvement"]
df["Performance"] = np.select(conditions, categories, default="Needs Improvement")

print("\n--- 11. Updated DataFrame with Result and Performance ---")
print(df)

# 12. Find the department having the highest average FinalMarks
highest_dept = df.groupby("Department")["FinalMarks"].mean().idxmax()
print("\n--- 12. Department with Highest Average FinalMarks ---")
print(highest_dept)
```

employees = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",

    ],

    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],

    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5
    ],

    "Salary": [
        60000, 45000, 75000, 65000,

    ]
})


Determine:

Average salary.
Maximum and minimum salary.
Employee receiving the highest salary.
Employees earning above ₹60,000.
Average salary for each department.
Average experience by department.
Number of employees in each department.
Sort employees by salary.
Find IT employees with more than 3 years of experience.
Add a SalaryLevel column:
	≥ ₹75,000 → High
	₹50,000–₹74,999 → Medium
	< ₹50,000 → Low

```python
import numpy as np
import pandas as pd

# Creating the employees DataFrame
employees = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",
        "E05", "E06", "E07", "E08"
    ],
    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],
    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5
    ],
    "Salary": [
        60000, 45000, 75000, 65000,
        55000, 85000, 48000, 70000
    ]
})

# 1. Average salary
print("--- 1. Average Salary ---")
print(employees["Salary"].mean())

# 2. Maximum and minimum salary
print("\n--- 2. Maximum and Minimum Salary ---")
print(f"Maximum Salary: {employees['Salary'].max()}")
print(f"Minimum Salary: {employees['Salary'].min()}")

# 3. Employee receiving the highest salary
print("\n--- 3. Employee Receiving Highest Salary ---")
highest_earner = employees.loc[employees["Salary"].idxmax()]
print(highest_earner)

# 4. Employees earning above ₹60,000
print("\n--- 4. Employees Earning Above ₹60,000 ---")
print(employees[employees["Salary"] > 60000])

# 5. Average salary for each department
print("\n--- 5. Average Salary by Department ---")
print(employees.groupby("Department")["Salary"].mean())

# 6. Average experience by department
print("\n--- 6. Average Experience by Department ---")
print(employees.groupby("Department")["Experience"].mean())

# 7. Number of employees in each department
print("\n--- 7. Number of Employees in Each Department ---")
print(employees.groupby("Department")["Employee"].count())

# 8. Sort employees by salary (descending order)
print("\n--- 8. Employees Sorted by Salary (Descending) ---")
print(employees.sort_values(by="Salary", ascending=False))

# 9. Find IT employees with more than 3 years of experience
print("\n--- 9. IT Employees with > 3 Years Experience ---")
print(employees[(employees["Department"] == "IT") & (employees["Experience"] > 3)])

# 10. Add a SalaryLevel column (with string default to avoid type error)
conditions = [
    employees["Salary"] >= 75000,
    (employees["Salary"] >= 50000) & (employees["Salary"] < 75000),
    employees["Salary"] < 50000
]
levels = ["High", "Medium", "Low"]
employees["SalaryLevel"] = np.select(conditions, levels, default="Low")

print("\n--- 10. Updated DataFrame with SalaryLevel ---")
print(employees)
```
