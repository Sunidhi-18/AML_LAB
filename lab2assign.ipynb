5. Create an array containing numbers from 1 to 24. Reshape it into:
4 × 6
6 × 4
2 × 3 × 4
Display the shape and ndim of each result.

```python
arr = np.arange(1,25)
#4 × 6
arr1 = arr.reshape(4,6)
print("shape:", arr1.shape, "|ndim:", arr1.ndim)
print(arr1)
#6 x 4
arr2 = arr.reshape(6,4)
print("shape:", arr2.shape, "|ndim:", arr2.ndim)
print(arr2)
#2 × 3 × 4
arr3 = arr.reshape(2,3,4)
print("shape:", arr3.shape, "|ndim:", arr3.ndim)
print(arr3)
```


8. Given:
salary = np.array([25000, 32000, 28000, 45000,
                   50000, 38000, 42000, 30000])

Find:
Total salary
Average salary
Highest salary
Lowest salary
Median salary
Standard deviation

```python
salary = np.array([25000, 32000, 28000, 45000, 50000, 38000, 42000, 30000])

print("total salary = ", np.sum(salary))
print("avg salary = ", np.mean(salary))
print("highest salary = ", np.max(salary))
print("lowest salary = ", np.min(salary))
print("median salary = ", np.median(salary))
print("std.dev of salary = ", np.std(salary))
```

11. Create:

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

Perform:

Matrix addition
Element-wise multiplication
Matrix multiplication
Transpose of A.

explain the difference between:
A * B and A @ B

```python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

# Operations
add_res = A + B
elem_mult = A * B
matmul_res = A @ B  # or np.dot(A, B)
trans_A = A.T

print("Matrix Addition (A + B):\n", add_res)
print("\nElement-wise Multiplication (A * B):\n", elem_mult)
print("\nMatrix Multiplication (A @ B):\n", matmul_res)
print("\nTranspose of A (A.T):\n", trans_A)
```
