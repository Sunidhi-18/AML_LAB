```python
#matrix creation by array input

num= np.matrix([[11,121,130],
[613,34,13],
[514,50,16]])
num
```

```python
print("main diagonal elements:\n", np.diag(num))
print("\nDiagonal elements above main diagonal:\n", np.diag(num,1))
print("\nDiagonal elements below main diagonal:\n", np.diag(num,-1))
```

```python
#operations
#use +,-,*,/ and ** to perform element wise addition, sub, multi, division
x= np.array([1,2,3])
y= np.array([4,5,6])
print(x)
print(y)
```

```python
print(x+y) #elementwise addition
print(x-y) #elementwise subtration
print(x*y) #elementwise multiplication
print(x/y) #elementwise division
print(x**y) #elementwise power
print(x@y) #dot product
```

```python
x.dot(y) #dot product
```
